```python
import pandas as pd
import numpy as np
import torch

from transformers import AutoTokenizer
from transformers import AutoModelForSequenceClassification
```

```python
import zipfile

with zipfile.ZipFile('/content/train.csv.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/')
print('train.csv extracted successfully.')
```

```python
df = pd.read_csv("/content/train.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
df = pd.read_csv("/content/train.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
labels = [
    "toxic",
    "severe_toxic",
    "obscene",
    "threat",
    "insult",
    "identity_hate"
]

print("Target labels:")
print(labels)

print("\nLabel counts:")
print(df[labels].sum())
```

```python
sample_text = "You are stupid and I hate you."

print("Sample comment:")
print(sample_text)
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased"
)

print("Tokenizer loaded successfully.")
```

```python
tokens = tokenizer.tokenize(sample_text)

print("Original text:")
print(sample_text)

print("\nTokens:")
print(tokens)
```

```python
encoded_input = tokenizer(
    sample_text,
    padding="max_length",
    truncation=True,
    max_length=32,
    return_tensors="pt"
)

print("Encoded input created successfully.")
```

```python
print("Input IDs:")
print(encoded_input["input_ids"])

print("\nInput IDs shape:")
print(encoded_input["input_ids"].shape)
```

```python
print("Attention Mask:")
print(encoded_input["attention_mask"])

print("\nAttention Mask shape:")
print(encoded_input["attention_mask"].shape)
```

```python
model = AutoModelForSequenceClassification.from_pretrained(
    "distilbert-base-uncased",
    num_labels=6,
    problem_type="multi_label_classification"
)

print("DistilBERT model loaded successfully.")
```

```python
print(model)
```

```python
model.eval()

with torch.no_grad():
    outputs = model(**encoded_input)

print("Raw logits:")
print(outputs.logits)
```

```python
probabilities = torch.sigmoid(outputs.logits)

print("Probability scores:")
print(probabilities)
```

```python
prediction_results = {
    label: float(score)
    for label, score in zip(labels, probabilities[0])
}

for label, score in prediction_results.items():
    print(f"{label}: {score:.4f}")
```

```python
def predict_toxicity(text):
    encoded = tokenizer(
        text,
        padding="max_length",
        truncation=True,
        max_length=32,
        return_tensors="pt"
    )

    model.eval()

    with torch.no_grad():
        output = model(**encoded)

    scores = torch.sigmoid(output.logits)[0]

    results = {
        label: round(float(score), 4)
        for label, score in zip(labels, scores)
    }

    return results
```

```python
test_comment = "You are stupid and I hate you."

results = predict_toxicity(test_comment)

print("Input comment:")
print(test_comment)

print("\nPrediction results:")
for label, score in results.items():
    print(f"{label}: {score}")
```
